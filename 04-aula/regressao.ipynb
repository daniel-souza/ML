```python
import pandas as pd
diabetes_df = pd.read_csv("diabetes_clean.csv")
```

```python
# Limpeza de dados
diabetes_df = diabetes_df[
    (diabetes_df['bmi'] > 0) &
    (diabetes_df['glucose'] > 0) &
    (diabetes_df['diastolic'] > 0)
]
```

```python
diabetes_df.head(10)
```

```python
X = diabetes_df.drop("glucose", axis=1).values
y = diabetes_df["glucose"].values
print(type(X), type(y))
```

```python
X_bmi = X[:, 4]
print(y.shape, X_bmi.shape)
```

```python
X_diastolic = X[:,1]
X_diastolic
```

```python
X_bmi = X_bmi.reshape(-1, 1)  # cria uma matriz 2D de uma única coluna.
X_diastolic = X_diastolic.reshape(-1, 1)
print(y.shape, X_bmi.shape)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X_bmi, y)
plt.ylabel("Blood Glucose (mg/dl)")
plt.xlabel("Body Mass Index")

plt.show()
```

```python
plt.scatter(X_diastolic, y)
plt.ylabel("Blood Glucose (mg/dl)")
plt.xlabel("Diastolic")

plt.show()
```

```python
X_dpf = diabetes_df['dpf'].values.reshape(-1, 1)
plt.scatter(X_dpf, y)
plt.ylabel("Blood Glucose (mg/dl)")
plt.xlabel("DPF")

plt.show()
```
